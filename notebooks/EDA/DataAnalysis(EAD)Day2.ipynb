# DataSet

## Generate Data using numpy

```python
# khud dy data set bnayn gy our usko csv file ma convert kr k again usko use karen gy sy analysis apply karen gy 

names = np.char.add('std', np.arange(1,101).astype('str'))
```

```python
ages  = np.random.randint(18,40,100)
ages[:10]
```

```python
gender = np.random.choice(['Male','Female'],100)
cnic_p1 = np.random.randint(10000, 100000, 100).astype(str)
cnic_p2 = np.random.randint(100000, 1000000, 100).astype(str)
cnic_p3 = np.random.randint(0, 10, 100).astype(str)

cnic_p1 = np.char.add(cnic_p1,'-')
cnic_p2 = np.char.add(cnic_p1,cnic_p2)
cnic_p3 = np.char.add('-',cnic_p3)

cnic = np.char.add(cnic_p2,cnic_p3)



ml_100 = np.random.randint(20,101,100)
dl_100 = np.random.randint(10,101,100)
nlp_100 = np.random.randint(40,101,100)
eda_60 = np.random.randint(0,61,100)

```

## Introduce Null values in Data

```python
cnic= cnic.astype(object)

cnic[np.random.randint(0,100,5)]=np.nan
```

## convert it into Dictionary

```python
data  = {
    'Name':names,
    'Ages':ages,
    'Gender':gender,
    'CNIC':cnic,
    'Machine Learning': ml_100,
    'Deep Learning': dl_100,
    'NLP': nlp_100,
    'EDA': eda_60
}
```

## convert it into DataFrames

```python
df=pd.DataFrame(data)
df.head(1)
```

## Saving as CVS file

```python
df.to_csv("Student_Records.csv", index=False)
```

## Feature Engineering

```python
df['Obtained Marks']=df['Machine Learning']+df['Deep Learning']+df['NLP']+df['EDA']
df.head(2)
```

```python
df['Percentage']=df['Machine Learning']+df['Deep Learning']+df['NLP']+df['EDA']
df.head(2)
```

```python
df['Percentage'] = (df['Obtained Marks'] / 360 * 100).round(2)

df.head(3)
```

```python
df['Average'] = (df['Obtained Marks'] / 4).round(2)
df.head(2)
```

```python
#method 1 thorugh lemda
df['Status'] = df[ 'Percentage']. apply(lambda x: 'Passed' if x > 50 else 'Failed')
df.head(1)
```

```python
#method 2 thorugh func
```

```python
def check_Staus(x):
    if (x>50):
        return "Passed"
    else:
        return "Failed"


df['Status'] = df['Percentage'].apply(check_Staus)
```
