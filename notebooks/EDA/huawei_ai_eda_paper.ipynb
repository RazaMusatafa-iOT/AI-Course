**Q1. What is AutoML? Which tools provide it?**

AutoML means Automated Machine Learning. The system does the hard steps by itself like choosing the algorithm, tuning the settings and training the model, so we write little or no code.

**Q3. What is model evaluation and why before deployment?**

Model evaluation means testing the trained model on new test data that it has not seen, using metrics like accuracy (classification) or R2 and MAE (regression).
We do it before deployment to make sure the model works on real data, to catch overfitting and to compare models, so only a good model reaches the users.

Training is when the model learns from data. Inference is when we use the trained model to predict on new data. Training takes a lot of time and is done once, but inference is fast and is used again and again.

**Q10. A city wants to monitor traffic using roadside cameras. Describe a solution that uses HiLens, EI and OBS. Show the data flow with a simple diagram (arrows are enough) and mention one challenge.**

The city puts smart cameras on busy roads to check if bike riders are wearing helmets. Each camera uses HiLens to run my helmet detection model on the camera itself, so it can find a rider without a helmet right away. It can also count the traffic.

The camera saves the photos of riders without helmets and the traffic counts in OBS. Then these photos go to Cloud EI, where OCR reads the bike number plate. The results are shown on a dashboard, so the traffic police can send a fine.

Camera -> HiLens (detect no helmet) -> OBS (save photos) -> Cloud EI (read number plate) -> Dashboard / Fine

Challenge: At night or in rain and fog the camera cannot see clearly, so it can miss a helmet or read the wrong number plate. Internet on the road can also be slow.

**Q11. Define MLOps in your own words and give two tasks that it automates.**

MLOps means taking care of a model after we make it, in an automatic way. It is like DevOps but for machine learning. For example, in my helmet project MLOps can:
1. train the helmet model again by itself when new helmet photos come
2. put the new model version live without doing it by hand

**Q12. What is the difference between a metric and an alarm in Cloud Eye?**

A metric is a number that Cloud Eye keeps checking, like CPU usage or how fast my helmet model replies.
An alarm is a rule on that number. When it goes over a limit, it informs us. For example: "if the helmet model takes more than 1 second, send me an email".

**Q15. Write a monitoring and automation plan for an AI service deployed on Huawei Cloud. Cover what you would monitor, which alarms you would set, who is notified, and which automated responses you would add.**

My service is the helmet detection model on Huawei Cloud.

**Monitor (Cloud Eye):** CPU usage, reply time, errors, and model accuracy.

**Alarms:** reply time above 1 second, errors above 5%, CPU above 80%, or accuracy below 90%.

**Notified:** the engineer gets SMS/email for slow or down service, and the ML team for low accuracy.
