```python

#Task

students = ['Ali', 'Ahmed']
math    = np.array([7, 4])
physics = np.array([8, 6])
comp    = np.array([6, 9])
x = np.arange(len(students))   # [0, 1]
w = 0.25 

fig, ax = plt.subplots(1, 2, figsize=(12, 5))

# Grouped
for offset, marks, name, c in [(-w, math, 'Math', 'steelblue'),
                               (0, physics, 'Physics', 'orange'),
                               (w, comp, 'Comp', 'green')]:
    bars = ax[0].bar(x + offset, marks, w, label=name, color=c)
    ax[0].bar_label(bars)
ax[0].set_xticks(x, students)
ax[0].set_ylim(0, 10)
ax[0].set_title('Raza Mustafa')
ax[0].legend()

# Stacked
# Stacked
s1 = ax[1].bar(students, math, label='Math', color='steelblue')
s2 = ax[1].bar(students, physics, bottom=math, label='Physics', color='orange')
s3 = ax[1].bar(students, comp, bottom=math + physics, label='Comp', color='green')

# Har hisse ke andar number
ax[1].bar_label(s1, label_type='center', color='white', fontweight='bold')
ax[1].bar_label(s2, label_type='center', color='white', fontweight='bold')
ax[1].bar_label(s3, label_type='center', color='white', fontweight='bold')

# Top pe total
total = math + physics + comp
for i, t in enumerate(total):
     ax[1].text(i, t + 0.5, t, ha='center', fontweight='bold')

ax[1].set_ylim(0, 30)
ax[1].set_title('Raza Mustafa')
ax[1].legend()

plt.savefig('StudentsData.jpg')
plt.show()

```

# Building EDA Report using Subplot

```python
fig, ax = plt.subplots(2, 3, figsize=(10, 4))
ax[0,0].plot(x,y)
ax[0,0].set_title('1st image')
ax[0,1].plot(y,x)
ax[0,1].set_title('2nd image')
ax[0,2].plot(x,y)
ax[0,2].set_title('3rd image')
ax[1,0].plot(y,x)
ax[1,0].set_title('4th image')
ax[1,1].plot(y,x)
ax[1,1].set_title('5th image')
ax[1,2].plot(y,x)
ax[1,2].set_title('6th image')
plt.tight_layout()
```

```python
fig, ax = plt.subplots(2, 2, figsize=(10, 6))

ax[0, 0].plot([1, 2, 3], [10, 20, 30])
ax[0, 0].set_title("Line Plot")

ax[0, 1].bar([1, 2, 3], [10, 20, 15])
ax[0, 1].set_title("Bar Plot")

ax[1, 0].scatter([1, 2, 3, 4], [5, 15, 10, 24])
ax[1, 0].set_title("Scatter Plot")

ax[1, 1].hist([10, 20, 20, 30, 30, 30])
ax[1, 1].set_title("Histogram")

plt.tight_layout()
plt.show()
```
