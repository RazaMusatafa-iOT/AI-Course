```python
#TASK
# 1. Create a 4x4 matrix with random integers from 1 to 100
matrix = np.random.randint(1, 101, size=(4, 4))
print("Original Matrix:")
print(matrix)

# 2. Find minimum and maximum
print("Minimum:", np.min(matrix))
print("Maximum:", np.max(matrix))

# 3. Find mean and standard deviation
print("Mean:", np.mean(matrix))
print("Standard Deviation:", np.std(matrix))

# 4. Reshape the matrix into a 1D array
arr_1d = matrix.reshape(16)
print("1D Array:")
print(arr_1d)
```
