```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('darkgrid')
```

## Fundamentals of Seaborn

```python
df = sns.load_dataset('tips')

df.head()
```

## Line plot

```python
#xlable ='Total bill'
#ylable ='Tip'
plt.figure(figsize= (20,4))
sns.lineplot(x='total_bill', y='tip', data=df)
df.head()
df.value_counts()
```

```python
plt.figure(figsize= (15,4))
sns.barplot(x='total_bill', y='tip', data=df)
df.head()
```

## Count Plot

```python
sns.countplot(x='day', data=df)
plt.title('Count days')
plt.savefig('Countplot.png')
df.head()
```

## Bar Plot

```python
sns.barplot(x='day',y ='total_bill', data=df)
plt.title('Count days')
plt.savefig('Barplot.png')
df.head()
```

## Box Plot

```python
sns.boxplot(x='day',y ='total_bill', data=df)
plt.title('days v total bill')
plt.savefig('Boxplot.png')
df.head()
```

## Violin Plot

```python
plt.figure(figsize= (15,4))
sns.violinplot(x='day',y ='total_bill', data=df)
plt.title('days v total bill')
plt.savefig('Violinplot.png')
df.head()
```

## Swarm Plot

```python
plt.figure(figsize= (15,4))
sns.swarmplot(x='day',y ='total_bill', data=df)
plt.title('days v total bill')
plt.savefig('swarmlot.png')
```

## Stripe Plot

```python
plt.figure(figsize= (15,4))
sns.stripplot(x='day',y ='total_bill', data=df, jitter = True)
plt.title('days v total bill')
plt.savefig('stripelot.png')
```

## Histograme Plot

```python
plt.figure(figsize= (15,4))
sns.histplot(df['total_bill'], bins= 30)
plt.title('Frequency Count of total bill')
plt.savefig('histoplot.png')
```

```python
plt.figure(figsize= (15,4))
sns.histplot(df['total_bill'], bins= 30, kde = True)
plt.title('Frequency Count of total bill')
plt.savefig('histo_KDE_plot.png')
```
