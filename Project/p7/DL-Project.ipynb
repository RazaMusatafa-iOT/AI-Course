# HandWritten Digit Classification using ANN
## Imports

```python
! pip install scikit-learn tensorflow

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import confusion_matrix



from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

## Load Dataset

```python
digits = load_digits()
```

```python
X = digits.data
y = digits.target
```

```python
fig,ax = plt.subplots(5,10)

for i in range(5):
    for j in range(10):
        ax[i,j].imshow(digits.images[i+1*j+1], cmap= 'gray')
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2)
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

## Scale and Encode

```python
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
y_train = to_categorical(y_train)
y_test = to_categorical(y_test)
```

## Model Developemnt

```python
model = Sequential()
```

```python
model.add(Dense(64, activation='relu', input_shape=(64,)))
```

```python
model.add(Dense(32, activation='relu'))
```

```python
model.add(Dense(10, activation='relu'))
```

```python
model.compile(optimizer='adam', metrics=['accuracy'],loss='categorical_crossentropy')
```

```python
history = model.fit(
    X_train, y_train,
    epochs=30,
    batch_size=16,
    validation_split=0.1
)
```

```python
model.summary()
```

```python
model.fit(
    X_train, y_train,       # tasveerein aur unke sahi jawab
    epochs=30,              # poori kitaab 30 dafa parho
    batch_size=16,          # ek waqt mein 16 tasveerein dekh ke wazan sudharo
    validation_split=0.1    # 10% alag rakho, har round ke baad mock test
)
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred = np.argmax(y_pred, axis=1)
```

```python
y_test =  np.argmax(y_test, axis=1)
```

```python
cm = confusion_matrix(y_test, y_pred)
```

```python
sns.heatmap(cm, annot=True)
```
