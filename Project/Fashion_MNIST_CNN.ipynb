## Step 1 : Import libraries and load the dataset

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
```

## Step : 1.2 Load Fashion MNIST dataset

```python
# Load Fashion MNIST dataset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()
```

## Check dataset dimensions

```python
print("Training Images:", x_train.shape)
print("Training Labels:", y_train.shape)

print("Testing Images:", x_test.shape)
print("Testing Labels:", y_test.shape)
```

## Display an image

```python
class_names = [
    'T-shirt/top',
    'Trouser',
    'Pullover',
    'Dress',
    'Coat',
    'Sandal',
    'Shirt',
    'Sneaker',
    'Bag',
    'Ankle boot'
]

plt.imshow(x_train[0], cmap='gray')
plt.title(class_names[y_train[0]])
plt.axis('off')
plt.show()
```

## Normalize training and testing images

```python
x_train = x_train / 255.0
x_test = x_test / 255.0
```

## Check current shape of images

```python
print(x_train.shape)
```

## Reshape training and testing images

```python
# Reshape training images
x_train = x_train.reshape(-1, 28, 28, 1)

# Reshape testing images
x_test = x_test.reshape(-1, 28, 28, 1)
```

## Verify preprocessing

```python
print("Training Shape:", x_train.shape)
print("Testing Shape:", x_test.shape)

print("Minimum Pixel Value:", x_train.min())
print("Maximum Pixel Value:", x_train.max())
```

## Display an image after preprocessing

```python
plt.imshow(x_train[0].squeeze(), cmap='gray')
plt.title(class_names[y_train[0]])
plt.axis('off')
plt.show()
```

## Initialize the CNN model

```python
model = tf.keras.Sequential()
```

## Define the input layer

```python
model.add(tf.keras.layers.Input(shape=(28, 28, 1)))
```

## Add the Convolution layer

```python
model.add(tf.keras.layers.Conv2D(
    filters=32,
    kernel_size=(3, 3),
    padding='same'
))
```

## Batch Normalization

```python
model.add(tf.keras.layers.BatchNormalization())
```

## Add ReLU activation

```python
model.add(tf.keras.layers.Activation('relu'))
```

## Add Max Pooling

```python
model.add(tf.keras.layers.MaxPooling2D(pool_size=(2, 2)))
```

## Complete code for the first CNN block

```python
# Initialize CNN model
model = tf.keras.Sequential()

# Input Layer
model.add(tf.keras.layers.Input(shape=(28, 28, 1)))

# Convolution Layer
model.add(tf.keras.layers.Conv2D(
    filters=32,
    kernel_size=(3, 3),
    padding='same'
))

# Batch Normalization
model.add(tf.keras.layers.BatchNormalization())

# ReLU Activation
model.add(tf.keras.layers.Activation('relu'))

# Max Pooling
model.add(tf.keras.layers.MaxPooling2D(pool_size=(2, 2)))
```

## Check Model

```python
model.summary()
```

## Add the Second CNN Block

```python
# Second CNN Block

model.add(tf.keras.layers.Conv2D(
    filters=64,
    kernel_size=(3, 3),
    padding='same'
))

model.add(tf.keras.layers.BatchNormalization())

model.add(tf.keras.layers.Activation('relu'))

model.add(tf.keras.layers.MaxPooling2D(pool_size=(2, 2)))
```

## Model Summary

```python
model.summary()
```

## Flatten Layer

```python
model.add(tf.keras.layers.Flatten())
```

## Model Summary

```python
model.summary()
```

## Dense Layer

```python
# Dense Layer
model.add(tf.keras.layers.Dense(128))
```

## Add Batch Normalization

```python
model.add(tf.keras.layers.BatchNormalization())
```

## Add ReLU Activation

```python
model.add(tf.keras.layers.Activation('relu'))
```

## Add Dropout

```python
model.add(tf.keras.layers.Dropout(0.3))
```

## Model Summary

```python
model.summary()
```

## Add the Output Layer

```python
model.add(tf.keras.layers.Dense(10, activation='softmax'))
```

## Model Summary

```python
model.summary()
```

## Compile the model

```python
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

## Set up callbacks (EarlyStopping and ReduceLROnPlateau)

```python
early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor='val_loss',
    patience=4,
    restore_best_weights=True
)

reduce_lr = tf.keras.callbacks.ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    patience=2,
    min_lr=0.00001
)
```

## Train The Model

```python
history = model.fit(
    x_train,
    y_train,
    epochs=25,
    batch_size=64,
    validation_split=0.1,
    callbacks=[early_stopping, reduce_lr]
)
```

## Check whether the model is still usable

```python
print("Model exists:", model is not None)
print("Completed epochs:", len(history.history['accuracy']))
print("Latest training accuracy:", history.history['accuracy'][-1])
```

## Check the training history

```python
print("Training Accuracy:", history.history['accuracy'])
print("Validation Accuracy:", history.history['val_accuracy'])

print("Training Loss:", history.history['loss'])
print("Validation Loss:", history.history['val_loss'])
```

## Visualize training results

```python
plt.figure(figsize=(12, 4))

# Accuracy graph
plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

# Loss graph
plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

plt.tight_layout()
plt.show()
```

## Evaluate your model on test data

```python
test_loss, test_accuracy = model.evaluate(x_test, y_test)

print("Test Loss:", test_loss)
print("Test Accuracy:", test_accuracy * 100, "%")
```

## Generate predictions

```python
predictions = model.predict(x_test)

predicted_labels = np.argmax(predictions, axis=1)

print("Predicted labels:", predicted_labels[:10])
print("Actual labels:", y_test[:10])
```

## Create your prediction visualization

```python
plt.figure(figsize=(15, 15))

for i in range(25):

    plt.subplot(5, 5, i + 1)

    plt.imshow(x_test[i].squeeze(), cmap='gray')

    predicted = predicted_labels[i]
    actual = y_test[i]

    color = 'green' if predicted == actual else 'red'

    plt.title(
        f"Pred: {class_names[predicted]}\nActual: {class_names[actual]}",
        color=color
    )

    plt.axis('off')

plt.tight_layout()
plt.savefig("Fashion_MNIST_Predictions.jpeg")
plt.show()
```

## Confusion Matrix

```python
from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns

cm = confusion_matrix(y_test, predicted_labels)

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=class_names, yticklabels=class_names)
plt.xlabel('Predicted Label')
plt.ylabel('Actual Label')
plt.title('Confusion Matrix')
plt.show()
```

## Classification Report

```python
print(classification_report(y_test, predicted_labels, target_names=class_names))
```

## Test the model on a single sample image

```python
index = 15   # change this number to test another image

image = x_test[index]
prediction = model.predict(image.reshape(1, 28, 28, 1))
predicted_class = np.argmax(prediction)

plt.imshow(image.squeeze(), cmap='gray')
plt.title(f"Predicted: {class_names[predicted_class]}  |  Actual: {class_names[y_test[index]]}")
plt.axis('off')
plt.show()

print("Confidence:", round(np.max(prediction) * 100, 2), "%")
```

## Show the wrong predictions

```python
wrong = np.where(predicted_labels != y_test)[0]
print("Total wrong predictions:", len(wrong))

plt.figure(figsize=(12, 6))
for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_test[wrong[i]].squeeze(), cmap='gray')
    plt.title(f"P: {class_names[predicted_labels[wrong[i]]]}\nA: {class_names[y_test[wrong[i]]]}",
              color='red', fontsize=9)
    plt.axis('off')
plt.tight_layout()
plt.show()
```

## Save the model

```python
model.save("fashion_mnist_cnn.keras")
print("Model saved")
```

## Conclusion
- A CNN with 2 convolution blocks (Conv2D + BatchNormalization + ReLU + MaxPooling), a Dense layer with Dropout and a Softmax output layer was built to classify 10 clothing categories.
- Images were normalized (0–1) and reshaped to (28, 28, 1) before training.
- EarlyStopping and ReduceLROnPlateau were used to stop training at the best point and avoid overfitting.
- The model reached about **92% accuracy** on the test data.
- In the confusion matrix, most mistakes happen between similar items like **Shirt, T-shirt/top, Pullover and Coat**, while Trouser, Bag, Sandal and Ankle boot are classified very well.
